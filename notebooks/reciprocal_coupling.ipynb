```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
A = np.array([
    [2, 1],
    [1, 2],
])

i = np.array([np.sqrt(3)/2, 1/2])
# i = 2 * np.array([np.sqrt(3/2), 1/2])
j = [-1/2, np.sqrt(3)/2]

i_ = A @ i
j_ = A @ j

plt.quiver(0, 0, i[0], i[1], angles='xy', scale_units='xy', scale=1, color='r')
plt.quiver(0, 0, j[0], j[1], angles='xy', scale_units='xy', scale=1)
plt.quiver(0, 0, i_[0], i_[1], angles='xy', scale_units='xy', scale=1, color='r')
plt.quiver(0, 0, j_[0], j_[1], angles='xy', scale_units='xy', scale=1)

plt.xlim(-1, 3)
plt.ylim(-1, 3)
plt.axhline(y=0, color='black', linewidth=1)
plt.axvline(x=0, color='black', linewidth=1)
plt.gca().set_aspect('equal')
plt.grid(True, axis='both', color='gray', linestyle='--', linewidth=0.5)
```
